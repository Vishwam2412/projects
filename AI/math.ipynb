```python
#vector


class Vector:
    def __init__(self,component):
        self.component = component
        self.dim = len(component)
    
    def __sum__(self,other):
        return [x+y  for x , y in zip(self.component,other.component)]

    def __sub__(self,other):
        return [x-y  for x , y in zip(self.component,other.component)]
    
    def magnitude(self):
        return sum(x**2 for x in self.component)**0.5

    def normalize(self):
        magnitude = self.__magnitude__()
        return [x/magnitude for x in self.component]
    
    def __dot__(self,other):
        return sum(x*y for x ,y in zip(self.component,other.component))

    def __scale__(self,a):
        return [x*a for x in self.component]
    
    def __gram_schmidt__(self,other):
        z = Vector(self.__normalize__())
        b = other.__normalize__()
        n =  z-(b.__scale__(other.__magnitude__()))
        return n.__normalize__()
    
    def __repr__(self):
        return f"Vector({self.component})"

    def is_linearly_independent(vectors):

        n = len(vectors)
        dim = len(vectors[0].components)
        mat = [x.component[:] for x in vectors]
        rows = [row[:] for row in  mat.rows]

        rank = 0
        for col in range(dim):
            pivot = None
            for row in range(n):
                if rows[row][col]>1e-10:
                    pivot = row
                    break
            if pivot == None:
                continue
            
            rows[rank] , rows[pivot] = rows[pivot] , rows[rank]

            for row in range(n):
                if row!=rank and rows[row][col] > 1e-10:
                    mag = rows[row][col]
                    rows[rank] = [rows[row][j] - mag*rows[rank][j] for j in range(dim)]
            rank+=1

        return rank==n


        def project(a,b):
            scaler = a.dot(b)/b.dot(b)
            return [scaler*x for x in b.component]
        
        def gram_schmidt(vectors):
            orthonormal = []
            for v in vectors:
                w = v
                for u in orthonormal:
                    proj = project(v,u)
                    w = w - proj
                    if(w.magnitude() < 1e-10):
                        continue
                    orthonormal.append(w)
            return orthonormal

        

        n = len(vectors)
        dim = len(vectors[0].component)
        mat = [x.component[:] for x in vectors]
        rows = [row[:] for row in mat.rows]
        rank = 0

        for col in range(dim):
            pivot = None
            for row in range(n):
                if(rows[row][col] >= 1e-10):
                    pivot = row
                    break
            if pivot==None:
                continue

            rows[rank] , rows[pivot] = rows[pivot] , rows[rank]
            scale = rows[rank][col]
            rows[rank] = [x/scale for x in rows[rank]]



            for row in range(n):
                if row!=rank and abs(rows[row][col])>1e-10:
                    factor = rows[row][col]
                    rows[row] = [(rows[row][j] - factor*rows[rank][j])
                        for j in range(dim)
                    ]
            rank+=1
        return rank==n

x = Vector([1,2])
y = Vector([2,3])

z = x-y
z
# print()
```
